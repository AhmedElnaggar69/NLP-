```python
visitedIndex = []
built = False
global_corpusInChunks = []
global_dict = {}
global_words = []
def buildDict(corpus):
    
    words = word_tokenize(corpus)
    corpusInChunks = []


    for index in range(len(words)-2) :
        corpusInChunks.append(words[index:index+3])

    dict = {}


    for index in range(len(words)-2) :
        # we need to get counts for middle words
        wordBefore = words[index]
        mid_word = words[index+1]
        wordAfter = words[index+2]

        key = (wordBefore, wordAfter)
        if key not in dict:
            dict[key] = {}
        dict[key][mid_word] = dict[key].get(mid_word, 0) + 1

    # get probs
    
    for entry , insideDict in dict.items():
        total = sum(insideDict.values())
        insideDict["sum"] = total
        
    for entry , insideDict in dict.items():
        total = insideDict["sum"]
        for key in insideDict:
            if key != "sum":
                insideDict[key] = insideDict[key] / total
    return words , corpusInChunks , dict
    
```

```python

def spinMe(changesNeeded , corpus):
    global built, global_corpusInChunks , global_dict , visitedIndex, global_words
    if not built:
        global_words , global_corpusInChunks,global_dict = buildDict(corpus)
        built=True

    # in case changes are more then our chunks
    changesNeeded = min(changesNeeded , len(global_corpusInChunks))

    
    for i in range(changesNeeded):
        index = random.randint(0,len(global_corpusInChunks)-1)
        while index in visitedIndex:
            index = random.randint(0,len(global_corpusInChunks)-1)
        visitedIndex.append(index)

        chunk = global_corpusInChunks[index]
        contextKey = (chunk[0] , chunk[2])

        wordsToPick = []
        weights = []


        if contextKey in global_dict:
            dict = global_dict[contextKey]
            for key , value in dict.items():
                if key != "sum":
                    wordsToPick.append(key)
                    weights.append(value)
            if wordsToPick:
                sample = random.choices(wordsToPick , weights=weights , k=1)[0]
                global_words[index+1] = sample
                global_corpusInChunks[index][1] = sample
    return global_corpusInChunks
```
