```python
import re

def makeCleanString(column):
    final = []
    for entry in column:
        finalEntry = ""
        if not pd.isna(entry):
            finalEntry = re.sub(r'[^a-zA-Z\s]', ' ', str(entry))
            finalEntry = re.sub(r'\s+', ' ', finalEntry).strip()
            final.append(finalEntry)
        else :
            final.append("")
    return final
```

```python
Descriptions = df["Description"]
Wikipedia = df["Wikipedia"]
Comments = df["Comments"]
Tags = df["Tags"]
lists = df["Lists"]

cleanDescriptions = makeCleanString(Descriptions)
cleanWikipedia = makeCleanString(Wikipedia)
cleanComments = makeCleanString(Comments)
cleanTags = makeCleanString(Tags)
cleanLists = makeCleanString(lists)

df["cleanDescription"] = cleanDescriptions
df["cleanWikipedia"] = cleanWikipedia
df["cleanComments"] = cleanComments
df["cleanTags"] = cleanTags
df["cleanlists"] = cleanLists
```

```python
cleanDescription = df["cleanDescription"]
print(cleanDescription[4])
```

```python
df = df.drop(columns=["Description" , "Wikipedia" ,"Comments", "Tags" , "Lists"])
```

```python
df["fullString"] =  (df["cleanDescription"] + " ") * 3 + " " + (df["cleanWikipedia"] + " ") * 2 + " " + (df["cleanlists"] + " ") * 3 + " " + df["cleanComments"] + " " + (df["cleanTags"] + " ") * 2
```

```python
cleanString = df["fullString"]
print(cleanString[6])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


Vectorizer = TfidfVectorizer(stop_words='english', max_df=0.85, min_df=2 ,sublinear_tf=True ,ngram_range=(1, 2))

tfidf_mat = Vectorizer.fit_transform(df["fullString"])
```

```python
index = 30000

similarites = cosine_similarity(tfidf_mat[index] , tfidf_mat).flatten()
```

```python
similarites
```

```python
top10 = np.argsort(similarites)[::-1][1:51]
print(top10)

print("movie name is " + df["Movie"][index])
print("similer movies are :")
for top in top10:
    print("movie :" + df["Movie"][top])
```
