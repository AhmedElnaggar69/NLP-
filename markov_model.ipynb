```python
import math
class Markov :
    corpus = """ """
    words = []
    def __init__(self, corpus):
        self.corpus = corpus
        self.words = corpus.split()

    def markov(self):
        words = self.corpus.split()
        trans = {}
        for i in range(len(words)-1):
            current = words[i]
            next_word = words[i+1]

            if current not in trans:
                trans[current] = {}

            if next_word not in trans[current]:
                trans[current][next_word] = 0

            trans[current][next_word] += 1
        probs = {}
        for word , next_words in trans.items():
            total = sum(next_words.values())

            probs[word] = {}

            for next_word in next_words:
                count = next_words[next_word]
                probs[word][next_word] = count / total

        return probs,trans


    def check_sentence(self,sentence):
        _ , trans = self.markov()
        vocab = set(self.words)
        v = len(vocab)
        test_words = sentence.split()
        log_prob = 0.0

        for i in range(len(test_words) - 1):
            current = test_words[i]
            next_word = test_words[i + 1]

            if current in trans:
                total = sum(trans[current].values())
                count = trans[current].get(next_word, 0)
            else:
                total = 0
                count = 0

            p = (count + 1) / (total + v)
            log_prob += math.log(p)
        average = log_prob / (len(test_words) - 1) 
        return average
    
```

```python
import random
markovObj1 = Markov(corpus)
probs ,trans = markovObj1.markov()

current = "the"
sentence = [current]

for _ in range(30):
    if current not in probs:
        break

    current = random.choices(
        list(probs[current].keys()),
        weights=probs[current].values()
    )[0]

    sentence.append(current)

print(" ".join(sentence))
```
