```python
import numpy as np
import string
```

```python
dict = {} # start of phrase

first_order = {}

second_order = {}
```

```python
def removePunc(s):
    return s.translate(s.maketrans('','',string.punctuation))
```

```python
with open("nieText.txt", encoding="utf-8") as f:
    nietzsche_text = f.read()
with open("schText.txt", encoding="utf-8") as f:
    schopenhauer_text = f.read()
```

```python
def add2dict(dict , key , value):
    if key not in dict:
        dict[key] = []
    dict[key].append(value)
```

```python
with open("schText.txt", encoding="utf-8") as f:
    lines = f.readlines()          

for line in lines:                 
    clean_text = removePunc(line.rstrip().lower())
    tokens = clean_text.split()
    T = len(tokens)
    for i in range(T):            
        t = tokens[i]
        if i == 0:
            dict[t] = dict.get(t, 0.) + 1
        else:
            t_1 = tokens[i-1]
            if i == T-1:
                add2dict(second_order, (t_1, t), 'END')
            if i == 1:
                add2dict(first_order, t_1, t)
            else:
                t_2 = tokens[i-2]
                add2dict(second_order, (t_2, t_1), t)
```

```python
initalTotal = sum(dict.values())
for t,c in dict.items():
    dict[t] = c / initalTotal
```

```python
def list2pdict(ts):
    d = {}
    n = len(ts)

    for t in ts:
        d[t] = d.get(t,0.) + 1
    for t,c in d.items():
        d[t] = c/n
    return d
```

```python
for t_l , ts in first_order.items():
    first_order[t_l] = list2pdict(ts)
```

```python
for k , ts in second_order.items():
    second_order[k] = list2pdict(ts)
```

```python
def sample_word(d):
    p0 = np.random.random()

    cum = 0

    for t ,p in d.items():
        cum +=p
        if p0 < cum:
            return t
    assert(False)
```

```python
def generate():
    for i in range(3):
        sentence = []

        w0 = sample_word(dict)

        sentence.append(w0)

        w1 = sample_word(first_order[w0])
        sentence.append(w1)

        while(True):
            w2 = sample_word(second_order[(w0 , w1)])
            if w2 == 'END':
                break
            sentence.append(w2)
            w0 = w1
            w1 = w2
        print(' '.join(sentence))
```

```python
generate()
```
